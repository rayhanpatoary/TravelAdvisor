```python
from openai import OpenAI
from dotenv import load_dotenv

load_dotenv()

client = OpenAI()
```

```python
response = client.responses.create(
    model="gpt-4.1-mini",
    input="Hello, I want to plan a trip to Japan."
)

print(response.output_text)
```

```python
instructions = """
You are a friendly professional travel advisor.

Your job is to greet the user and collect the information needed to plan the trip
ask follow-up questions until it has gathered at least the following information:
- Destination(s)
- Trip duration
- interests (e.g., food, museums, hiking, beaches, nightlife, shopping)
avoid asking unnecessary questions once you have enough information
"""

response = client.responses.create(
    model="gpt-4.1-mini",
    instructions=instructions,
    input="Hello, I want to plan a trip to Japan."
)

print(response.output_text)
```
